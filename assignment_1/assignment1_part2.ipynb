```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

from sklearn.preprocessing import scale
from sklearn.linear_model import SGDRegressor
import sklearn.linear_model as skl_lm
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
import statsmodels.api as sm
import statsmodels.formula.api as smf
```

```python
advertising = pd.read_csv('Advertising.csv', usecols=[1,2,3,4])
advertising_X = advertising.loc[:,['TV','Radio','Newspaper']]
advertising_Y = advertising.loc[:,['Sales']]
```

```python
scaler = StandardScaler()
advertising_x_scaled = scaler.fit_transform(advertising_X)
```

```python
model = SGDRegressor(loss='squared_error',penalty=None, tol = 0,max_iter=10000,validation_fraction=0.1,early_stopping=True,n_iter_no_change= 5)
model.fit(advertising_X, advertising_Y.values.ravel())

print("--- Original Feature Data ---")
print('Intercept:', model.intercept_)
print('Coefficients:', model.coef_)
print('R2 Score:', model.score(advertising_X, advertising_Y))
```

```python
model_scaled = SGDRegressor(loss='squared_error',penalty=None, tol = 0,max_iter=10000,validation_fraction=0.1,early_stopping=True,n_iter_no_change= 5)
model_scaled.fit(advertising_x_scaled, advertising_Y.values.ravel())

print("--- Scaled Feature Data ---")
print('Intercept:', model_scaled.intercept_)
print('Coefficients:', model_scaled.coef_)
print('R2 Score:', model_scaled.score(advertising_x_scaled, advertising_Y))
```
