```python
import sklearn
import pandas as pd
import statsmodels.api as sm
```

```python
# categorical conversions
categorical_conversion = {
    "Good":0,
    "Medium":1,
    "Bad":2
}
yes_no_conversion = {
    "Yes": 1,
    "No": 0
}
df = pd.read_csv("Carseats.csv")
df["ShelveLoc"] = df["ShelveLoc"].map(categorical_conversion)
df["Urban"] = df["Urban"].map(yes_no_conversion)
df["US"] = df["US"].map(yes_no_conversion)
```

```python
X = df[["Price", "US"]]
X = sm.add_constant(X)
Y = df["Sales"]
```

```python
model = sm.OLS(Y, X).fit()
print(model.summary())
```
