## 🎤 Section 1 — Jennifer Lim (~7 min)
### Introduction & Theory

**Opening (~30 sec)**
- *"Hi everyone, we're Group 11 — I'm Jennifer, and I'm here with Palwinder and Jonathan. Today we're presenting LASSO Regression, which stands for Least Absolute Shrinkage and Selection Operator."*
- Briefly preview the agenda: theory → implementation → results.

**Step 1: Start with Linear Regression (~1 min)**
- LASSO is built on top of Linear Regression, so let's quickly recap that first.
- Linear Regression is a **supervised learning** algorithm that fits a straight line to predict a continuous target.
- Equation: $\hat{y} = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \dots + \beta_n x_n$.
- It's great for trend analysis, but it has two big weaknesses…

**Step 2: Problems with plain Linear Regression (~1 min)**
- **Overfitting:** with many features, it fits the noise instead of the signal.
- **No feature selection:** every feature gets a non-zero coefficient, even useless ones.
- This is exactly the gap LASSO fills.

**Step 3: Introduce LASSO (~2 min)**
- LASSO is Linear Regression **+ L1 regularization**.
- L1 means we add a penalty equal to the **absolute value of the coefficients** to the cost function.
- Show the cost function:
  $$ J(\beta) = \frac{1}{2m}\sum (\hat{y}_i - y_i)^2 + \lambda \sum |\beta_j| $$
- Walk through the symbols: $\hat{y}_i$ predicted value, $y_i$ true target, $\beta_j$ coefficient for feature *j*, $\lambda$ controls the strength of the penalty.

**Step 4: L1 vs L2 — why LASSO is special (~1.5 min)**
- Run the L1 vs L2 plot in the code notebook.
- L2 (Ridge) shrinks coefficients **toward** zero but never **to** zero.
- L1 (LASSO) can shrink coefficients **exactly to zero** → automatic feature selection.
- That's the key insight: LASSO doesn't just regularize, it **picks features for you**.

**Step 5: Wrap up theory (~30 sec)**
- Recap the three problems LASSO solves: overfitting, feature selection, interpretability.

**Handoff (~15 sec)**
> *"Now that we understand the theory, Palwinder will show how we actually build a LASSO model from scratch."*

## 🎤 Section 2 — Palwinder Singh (~7 min)
### Implementation — Building LASSO from Scratch

**Intro (~30 sec)**
- *"Thanks Jen. I'll walk through how we built our own LASSO regression class from scratch using gradient descent."*
- Mention the 4-step roadmap: imports → data prep → class definition → training.

**Step 1: Imports (~30 sec)**
- We use NumPy for math, pandas for data, scikit-learn for the dataset and the train/test split, and matplotlib for plots.
- We're **not** using sklearn's Lasso — we built our own.

**Step 2: Load and prep the data (~1.5 min)**
- Dataset: **California Housing** — 8 numerical features, target = median house value.
- Show the head of the dataframe.
- **Why we standardize:** the L1 penalty treats every coefficient the same, so features need to be on the same scale. Otherwise a feature with naturally large values would get unfairly penalized.
- 80/20 train/test split.

**Step 3: Walk through the LASSO class (~3 min)** — most important part
- `__init__`: stores learning rate, λ (lambda_param), and number of iterations.
- `fit()` is the heart of it:
  - Initialize weights to zero, bias to zero.
  - Loop for `n_iterations`:
    - Compute predictions: $\hat{y} = X\beta + b$.
    - Compute the error: $\hat{y} - y$.
    - Gradient w.r.t. weights = standard MSE gradient **+ λ · sign(β)** ← this is the L1 sub-gradient.
    - Update weights and bias by subtracting learning_rate × gradient.
- `predict()`: just $X\beta + b$.
- Emphasize: the only thing that makes this LASSO and not plain linear regression is that one extra `λ · sign(weights)` term.

**Step 4: Train and show convergence (~1.5 min)**
- Run the training cell live.
- Show the loss-curve plot — point out it drops fast and then flattens out → model has converged.
- Print the final loss and bias.

**Handoff (~15 sec)**
> *"The model has converged. Now Jonathan will show how well it actually performs and where LASSO shines in the real world."*

## 🎤 Section 3 — Jonathan Elahee (~6 min)
### Evaluation, Results & Real-World Applications

**Intro (~30 sec)**
- *"Thanks Palwinder. Now let's see how well the model actually performs on data it has never seen."*

**Step 1: Predictions & metrics (~1.5 min)**
- Generate predictions on the test set.
- Walk through each metric:
  - **R² (test):** how much of the variance the model explains. Higher is better, max = 1.
  - **R² (train):** for comparison — if it's much higher than test R², we're overfitting.
  - **MSE:** average squared error — heavier penalty for big mistakes.
  - **MAE:** average absolute error — more interpretable, in the same units as the target.

**Step 2: Feature selection (~1.5 min)**
- Show the ranked-coefficients table and bar chart.
- Point out which features got the biggest weights and which got pushed near zero.
- *"This is the magic of LASSO — it told us which features actually matter, automatically."*

**Step 3: Predicted vs actual scatter (~1 min)**
- Run the scatter plot.
- A tight cluster around the dashed 45° line means the model is doing well.
- Point out the upper boundary at $5 — that's a known cap in the dataset, not a model bug.

**Step 4: Regularization path (~1 min)**
- Show the λ sweep plot.
- As λ grows, more coefficients are pulled toward zero — visual proof of what Jennifer described in the theory.

**Step 5: Real-world applications (~45 sec)**
- **Finance:** predicting stock returns from many indicators.
- **Genomics:** picking a few relevant genes out of thousands.
- **Real estate:** literally what we just did.
- **Healthcare:** identifying the few variables that drive a patient outcome.
- LASSO shines when there are many features but only a few that actually matter.
